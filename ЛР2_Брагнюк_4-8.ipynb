## 1. Імпорт бібліотеки NumPy

```python
import numpy as np

print("NumPy version:", np.__version__)
```

## 2. Завдання: основні атрибути масиву

Створити матрицю розміру `(3, 4)` та перевірити її `shape`, `ndim`, `size`, `dtype` і `nbytes`.

```python
A = np.array([
    [6, 6, 6, 6],
    [3, 7, 9, 1],
    [10, 10, 10, 10]
], dtype=np.int64)

print("Матриця A:")
print(A)
print("\nshape:", A.shape)
print("ndim:", A.ndim)
print("size:", A.size)
print("dtype:", A.dtype)
print("itemsize:", A.itemsize)
print("nbytes:", A.nbytes)
```

**Висновок:** матриця має 3 рядки та 4 стовпці, тобто 12 елементів. `ndim = 2` означає, що масив двовимірний.

## 3. Завдання: перетворення типів даних

Створити масив `float64`, перетворити його на `float32` та порівняти `itemsize` і `nbytes`.

```python
a64 = np.array([1, 2, 3], dtype=np.float64)
a32 = a64.astype(np.float32)

print("a64:", a64)
print("dtype:", a64.dtype)
print("itemsize:", a64.itemsize)
print("nbytes:", a64.nbytes)

print("\na32:", a32)
print("dtype:", a32.dtype)
print("itemsize:", a32.itemsize)
print("nbytes:", a32.nbytes)
```

**Висновок:** один елемент `float64` займає 8 байтів, а `float32` — 4 байти. Тому після перетворення загальний обсяг пам'яті для цього масиву зменшується вдвічі.

## 4. Завдання: булева індексація

Із масиву `scores` вибрати лише значення від 70 до 90 включно.

```python
scores = np.array([56, 82, 91, 47, 73])

mask = (scores >= 70) & (scores <= 90)
selected = scores[mask]

print("scores:", scores)
print("mask:", mask)
print("Значення від 70 до 90:", selected)
```

## 5. Завдання: broadcasting

До кожного стовпця матриці додати різне число за допомогою broadcasting.

```python
A = np.array([
    [6, 6, 6, 6],
    [3, 7, 9, 1],
    [10, 10, 10, 10]
])

add_values = np.array([1, 2, 2, 3])
B = A + add_values

print("Початкова матриця:")
print(A)
print("\nЗначення для стовпців:", add_values)
print("\nРезультат:")
print(B)
```

**Пояснення:** вектор із 4 значень автоматично поширюється на кожний рядок матриці `(3, 4)`. Це і є механізм broadcasting.

## 6. Завдання: генерація випадкових даних

Потрібно згенерувати:
- матрицю `(4, 3)` із нормального розподілу;
- 10 випадкових цілих чисел від 1 до 100;
- вибірку з 5 **різних** індексів із діапазону `0..19`.

Для сучасного коду використовується `np.random.default_rng()`.

```python
rng = np.random.default_rng(seed=42)

normal_matrix = rng.normal(size=(4, 3))
random_integers = rng.integers(1, 101, size=10)
random_indices = rng.choice(np.arange(20), size=5, replace=False)

print("Матриця (4, 3) з нормального розподілу:")
print(normal_matrix)

print("\n10 випадкових цілих чисел від 1 до 100:")
print(random_integers)

print("\n5 різних індексів з діапазону 0..19:")
print(random_indices)
print("Усі індекси різні:", len(np.unique(random_indices)) == 5)
```

# 7. Підсумкове тренувальне завдання

Необхідно:
1. створити матрицю `X` форми `(5, 4)` із випадкових чисел;
2. обчислити середнє та стандартне відхилення кожної ознаки;
3. стандартизувати матрицю через broadcasting;
4. вибрати рядки, у яких значення першої ознаки додатне;
5. додати новий стовпець одиниць;
6. створити вектор ваг і обчислити прогноз `X @ w`;
7. для заданого `y_true` обчислити MSE;
8. перевірити масиви на `NaN` та `inf`.

```python
rng = np.random.default_rng(seed=42)

# 1. Матриця X форми (5, 4)
X = rng.normal(size=(5, 4))
print("1. Матриця X:")
print(X)
print("shape:", X.shape)
```

```python
# 2. Середнє та стандартне відхилення кожної ознаки
mean = X.mean(axis=0)
std = X.std(axis=0)

print("2. Середні значення:")
print(mean)
print("\nСтандартні відхилення:")
print(std)
```

```python
# 3. Стандартизація через broadcasting
X_scaled = (X - mean) / std

print("3. Стандартизована матриця:")
print(X_scaled)
print("\nСередні після стандартизації:")
print(X_scaled.mean(axis=0))
print("Стандартні відхилення після стандартизації:")
print(X_scaled.std(axis=0))
```

```python
# 4. Вибір рядків із додатним значенням першої ознаки
X_pos = X_scaled[X_scaled[:, 0] > 0]

print("4. Рядки з додатною першою ознакою:")
print(X_pos)
```

```python
# 5. Додавання стовпця одиниць
ones = np.ones((X_pos.shape[0], 1))
X_with_bias = np.hstack([ones, X_pos])

print("5. Матриця зі стовпцем одиниць:")
print(X_with_bias)
print("shape:", X_with_bias.shape)
```

```python
# 6. Вектор ваг і прогноз
w = rng.normal(size=X_with_bias.shape[1])
y_pred = X_with_bias @ w

print("6. Вектор ваг w:")
print(w)
print("\nПрогноз y_pred:")
print(y_pred)
```

```python
# 7. MSE
# Заданий вектор фактичних значень тієї ж довжини, що й прогноз
y_true = np.linspace(1.0, 2.0, num=X_pos.shape[0])
mse = np.mean((y_true - y_pred) ** 2)

print("7. y_true:", y_true)
print("y_pred:", y_pred)
print("MSE:", mse)
```

```python
# 8. Перевірка на NaN та inf
arrays_to_check = {
    "X": X,
    "X_scaled": X_scaled,
    "X_pos": X_pos,
    "X_with_bias": X_with_bias,
    "w": w,
    "y_pred": y_pred,
    "y_true": y_true
}

print("8. Перевірка масивів:")
for name, arr in arrays_to_check.items():
    has_nan = np.isnan(arr).any()
    has_inf = np.isinf(arr).any()
    print(f"{name}: NaN = {has_nan}, inf = {has_inf}")

all_finite = all(np.isfinite(arr).all() for arr in arrays_to_check.values())
print("\nУсі значення скінченні:", all_finite)
```

# 8. Контрольні запитання

**1. Чим `ndarray` відрізняється від списку Python?**  
`ndarray` — багатовимірний масив NumPy, елементи якого мають узгоджений тип `dtype`. Він підтримує швидкі векторизовані математичні операції без написання циклів Python.

**2. Що означають `shape`, `ndim`, `size`, `dtype`, `itemsize`?**  
- `shape` — розміри масиву за кожною віссю;
- `ndim` — кількість вимірів;
- `size` — загальна кількість елементів;
- `dtype` — тип елементів;
- `itemsize` — розмір одного елемента в байтах.

**3. У чому різниця між `np.arange` і `np.linspace`?**  
`np.arange(start, stop, step)` задає крок, а `np.linspace(start, stop, num)` — точну кількість рівновіддалених значень.

**4. Чим `view` відрізняється від `copy`?**  
`view` може використовувати ті самі дані, що й вихідний масив, тому його зміна може вплинути на оригінал. `copy` створює незалежну копію.

**5. Що таке векторизація?**  
Векторизація — виконання операцій одразу над масивами без явних Python-циклів.

**6. Які правила broadcasting?**  
NumPy порівнює форми масивів справа наліво. Розміри сумісні, якщо вони однакові або один із них дорівнює 1.

**7. Як трактувати `axis=0` і `axis=1`?**  
Для двовимірної матриці `axis=0` виконує агрегування вниз по рядках для кожного стовпця, а `axis=1` — уздовж стовпців для кожного рядка.

**8. Чим відрізняються `*` та `@`?**  
`*` виконує поелементне множення, а `@` — матричний добуток.

**9. Чому для `Ax=b` краще використовувати `np.linalg.solve`?**  
`np.linalg.solve(A, b)` безпосередньо розв'язує систему і не потребує явного обчислення оберненої матриці.

**10. Чому для нового коду рекомендовано `np.random.default_rng()`?**  
Це сучасний API генератора випадкових чисел NumPy. Він дає окремий об'єкт генератора, зручне керування `seed` та не залежить від глобального стану старого API.
