## Завдання 2. Скалярний добуток і довжина векторів

Дано вектори:
```python
u = np.array([2, -3, 6])
v = np.array([5,  1, 4])
```
1. Обчисліть скалярний добуток двома способами: `np.dot()` і оператором `@`.
2. Знайдіть евклідову норму кожного вектора.
3. Обчисліть косинус кута між векторами.

````python
u = np.array([2, -3, 6])
v = np.array([5, 1, 4])

dot1 = np.dot(u, v)
dot2 = u @ v
norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)
cos_angle = dot1 / (norm_u * norm_v)

print("np.dot(u, v) =", dot1)
print("u @ v =", dot2)
print("||u|| =", norm_u)
print("||v|| =", norm_v)
print("cos =", cos_angle)
````

## Завдання 11. Перестановка рядків і стовпців

Дано матрицю та порядки перестановок:
```python
A = np.array([[11,12,13,14], [21,22,23,24], [31,32,33,34], [41,42,43,44]])
row_order = np.array([2, 0, 3, 1])
col_order = np.array([3, 1, 0, 2])
```
Переставте рядки, потім стовпці, а після цього за допомогою обернених перестановок відновіть початкову матрицю.

````python
A = np.array([[11,12,13,14], [21,22,23,24],
              [31,32,33,34], [41,42,43,44]])
row_order = np.array([2, 0, 3, 1])
col_order = np.array([3, 1, 0, 2])

rows = A[row_order, :]
permuted = rows[:, col_order]

inv_rows = np.argsort(row_order)
inv_cols = np.argsort(col_order)
restored = permuted[:, inv_cols][inv_rows, :]

print("Після перестановки рядків:\n", rows)
print("Після перестановки стовпців:\n", permuted)
print("Відновлена матриця:\n", restored)
print("Відновлення правильне:", np.array_equal(A, restored))
````

## Завдання 15. Комутатор двох матриць

Для кожної пари матриць обчисліть комутатор `[A,B] = AB - BA` та визначте, для якої пари виконується `AB = BA`. Для перевірки використайте `np.allclose()`.

````python
A1 = np.array([[3,0], [0,7]])
B1 = np.array([[2,0], [0,5]])
A2 = np.array([[0,2], [-1,0]])
B2 = np.array([[4,0], [0,1]])

comm1 = A1 @ B1 - B1 @ A1
comm2 = A2 @ B2 - B2 @ A2

print("Комутатор першої пари:\n", comm1)
print("AB = BA для першої пари:", np.allclose(A1 @ B1, B1 @ A1))
print("Комутатор другої пари:\n", comm2)
print("AB = BA для другої пари:", np.allclose(A2 @ B2, B2 @ A2))
````

## Завдання 16. Квадратична форма

Дано симетричну матрицю `A`, вектор `x` та матрицю векторів `X`. Обчисліть `xᵀAx`, а для всіх рядків `X` — квадратичні форми без циклу за допомогою `np.einsum()`.

````python
A = np.array([[5,1,0], [1,4,2], [0,2,3]])
x = np.array([2,-1,3])
X = np.array([[2,-1,3], [1,0,2], [-2,1,1]])

q = x @ A @ x
q_all = np.einsum('ij,jk,ik->i', X, A, X)

print("x^T A x =", q)
print("Квадратичні форми для X:", q_all)
````

## Завдання 18. Ранг матриці

Дано дві матриці:
```python
A = np.array([[2,1,0], [1,3,4], [0,2,5]])
B = np.array([[1,3,5], [2,6,10], [4,1,0]])
```

Знайдіть ранг кожної матриці та визначте, яка з них має повний ранг. Використайте `np.linalg.matrix_rank()`.

## Завдання 19. Власні значення та власні вектори

Дано симетричну матрицю:
```python
A = np.array([[6.0, 2.0], [2.0, 3.0]])
```
Знайдіть власні значення та власні вектори, перевірте `A @ V = V * λ` та виведіть результат через `np.allclose()`. Використайте `np.linalg.eigh()`.

````python
A = np.array([[6.0, 2.0], [2.0, 3.0]])

eigenvalues, V = np.linalg.eigh(A)
left = A @ V
right = V * eigenvalues

print("Власні значення:", eigenvalues)
print("Власні вектори:\n", V)
print("A @ V:\n", left)
print("V * λ:\n", right)
print("Перевірка:", np.allclose(left, right))
````

## Завдання 20. Метод найменших квадратів

Для точок
```python
x = np.array([0., 1., 2., 3., 4.])
y = np.array([2.2, 3.8, 6.1, 7.9, 10.2])
```
побудуйте пряму $y \approx a + bx$.

1. Сформуйте матрицю ознак зі стовпця одиниць і значень `x`.
2. Знайдіть коефіцієнти `a` і `b` методом найменших квадратів.
3. Обчисліть прогнозні значення `y`.
4. Знайдіть суму квадратів залишків.

Використайте `np.linalg.lstsq()`.

````python
x = np.array([0., 1., 2., 3., 4.])
y = np.array([2.2, 3.8, 6.1, 7.9, 10.2])

A = np.column_stack((np.ones_like(x), x))
coefficients, residuals, rank, s = np.linalg.lstsq(A, y, rcond=None)
a, b = coefficients
y_pred = A @ coefficients
sse = np.sum((y - y_pred) ** 2)

print("Матриця ознак:\n", A)
print(f"a = {a:.4f}")
print(f"b = {b:.4f}")
print("Прогнозні значення y:", np.round(y_pred, 4))
print(f"Сума квадратів залишків = {sse:.4f}")
````
